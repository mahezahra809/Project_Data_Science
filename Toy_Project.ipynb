```python
#Extracting input & output columns
#to fetch input columns only
#&fetch output columns only

#using iloc fetch columns -sir did
#x=df.iloc[:,0:2]
#Y=df.iloc[:,-1]

#using fancy indexing
#in a list simply pass the columns you need 
#to fetch column = df[[col1,col2]]

X=df[['cgpa','iq']]

#it is a 2d- 2 axis
```

```python
Y=df[['placement']]
```

```python
#scale the value - will do it after train test split, so that we can see what data goes in training & what in test ( we do not do
#like that but here we do) we can do any order 1)train test & then scale value or vice versa, here doing train test

```

```python
#train test split
#to do it we wil import sklearn library
#inside sklearn library we have a function train_test split
# in the function train_test split, we pass 3 parameters , (indep var, dep var,test_size) means kitna data test 
#from main data ,kitna we pass it here - means training data mai kitna % data jyega and test data mai kitna % data jyega.
#we mention 0.1 means - 100% of data , 10% will be in test data. & rest 90% will go in training data.

#IMP-
#In Scikit-learn library , train_test_split is a function, not a class. 

#sklearn - The top-level package(folder)of the scikit-learn library.
#model_selection - A subpackage(sub-folder ) containing model-selection tools.
#train_test_split -A function that divides data into training and testing sets.
#----------
#sklearn → top-level package
#model_selection → subpackage
#train_test_split → function

#---------
#It is used to divide your dataset into separate parts, usually training data and testing data.
#--from (sklearn.model_selection import train_test_split )

from sklearn.model_selection import train_test_split 
train_test_split(X,Y,test_size=0.1)
```

```python
#we can see it has done the split,
#input data =ko 90 ,10 mai alg kr diya & 
#output data= ko 90, 10 alg kr diya

#1 Dataframe - X of training set, ( input ) can see 90 rows ( total were 100)
#2 Dataframe - X of test set ( input), can see 10 rows 

#3 Dataframe - Y of training set ( output),  can see 90 rows ( total were 100)
#4 Dataframe - Y of test set ( output ), can see 10 rows

#All 4 dataframes we can store it in 4 variables 
# we have training and test set different now

#randomly koi bhi row uth kr traing mai chala gya ( we can see it in training input & outpu t, pura row hi gya, hai 
#y train &x train ka row ek hi hoga, usi ka output hai wo y train mai ( same idnex we can see - as row uth kr gya hai)
#same for training data ( row uth kr jrhi hai puri)


#So train test split we have done - means training data & test data we have seperated them

```

```python
X_train, X_test, Y_train,Y_test = train_test_split(X,Y,test_size=0.1)
```

```python
#now scaling the value form -1 to 1 
#( means sari values ko -1& 1 ke beech kr do),
#not mendate to do but it a good to do it
#To do scaling 

#scikit learn (library) inside it we have Sklearn (package) inside it we have another package (preprecessing ) and 
#inside it we have a class - StanddardScalar
#of this class we will make an object - Scalar
#to scale we will use scalar.fit_tranform( )
#fit_transform() will first understand the data, and then transform that data( in a range from -1 to 1)
#when we run x_train we will see that all values are tranformed now from -1 to 1


#so X_train data we have, those value are now in range form -1 to 1 
#earlier it was dataframe( table ) not it has become array, we cannot see which row it is we cant see the row index,
#we saw before 
#-------------------------------
#CONCEPT -
#4.26981266e-01   means 0.426981266
#Move the decimal point 1 place left.

# 94705880e-02 -->  Move the decimal point 2 places left. --> 0.019470588

#e-01 means multiply by \(10^{-1}), or divide by 10.
#e-02 means multiply by \(10^{-2}), or divide by 100.
#e+00 means multiply by 1.
#e+01 means multiply by 10.
#so our entire data value in x_train is between -1 &1 ( and around)
#---------------------------------
#Same will do for data X_test 
#------------



from sklearn.preprocessing import StandardScaler
scalar =StandardScaler( )
X_train= scalar.fit_transform(X_train)
X_train
```

```python
#will use transform here, coz pattern we have learned during training time
#Scaling is done

X_test= scalar.transform(X_test)
X_test
```

```python
#Now training the model
#to train model we have decided that we will use logistic regression
#inside sklearn library , linear_model , inside it we have logistic regresssion class
#CONCEPT
#linear_model -->  A module inside Scikit-learn.
#LogisticRegression-->	A class inside the linear_model module.

#will create a LogisiticRegression class object-clf
#now clf(object) has power to access those functions that are inside logistic regression class
#class import kri iss class ke andr function aur kyi chezein hoskti.
#there there is a function named fit we will use it, fit( ) we use to train the model

#we just need to provide our traning data.

#class ek.py file( module mai hi hoti hai)

from sklearn.linear_model import LogisticRegression
clf =LogisticRegression()

#this is model training
clf.fit(X_train,Y_train)

#model get trained the result we get in logistic regression praemeter is telling that model is trained
#as data we small it got trained so fast
```

```python
#Now will evaluate the model
#we do Model evaluation - we find the accuracy on test data( means that model is working fine or not we are testing)
#inside predict function we pass new data it , it will do prediction, ( means result output it will give on the input data we have given)
#so 10 data it has given output for the input we have given, 
#this is prediction of our model
#now we will see the input test column that we have taken out and we will match ( with y_test input)
#we can see the input & output result it has predicted 
#it has given 8 correct results out of 10
#ye ek ek se match kro 1st wala 1st array usmai output hai (serial vise hi hua hai)

Y_predict =clf.predict(X_test)

#result of array we stored them in y_predict
```

```python
Y_test
```

```python
#now we will check accuracy of our model
#sklean module we have metrix module inside it we have function accuracy_score, 
#it will give the accurany of our model( ki y_test and y_pred ko match krega and tell kitna shi btya
#both has same shape  Y_test,Y_predict  - we pass correct data and the prediction data result.
#both should have same shape,0 
#For accuracy_score()= both arrays should ideally be 1D arrays with the same number of elements.
#need to see result mai kya arha array to uspr kam krte.

#0.8  means 80% accurancy, our model giving 80% accurate results.

from sklearn.metrics import accuracy_score

accuracy_score(Y_test,Y_predict)

```
